```python
# Importing data analysis package:

import pandas as pd

# Importing dataset:

dataset = pd.read_stata(
    r"C:\Users\jonat\OneDrive\Desktop\Uni\Gerber et al 2008\GerberGreenLarimer_APSR_2008_social_pressure.dta",
    convert_categoricals=False
)

variables = [
    "voted",
    "treatment",
    "hh_size",
    "g2004",
    "g2002",
    "g2000",
    "p2004",
    "p2002",
    "p2000",
    "sex",
    "yob"
]

# Compile them into a descriptive statistics table:

descriptive_stats = dataset[variables].describe().T

summary_table = descriptive_stats.rename(columns={
    "count": "N",
    "mean": "Mean",
    "50%": "Median",
    "std": "Standard Deviation",
    "min": "Minimum",
    "max": "Maximum"
})

display(summary_table.round(2))

# Assess data quality in terms of missing values for important variables:

summary_table["Missing N"] = dataset[variables].isna().sum()
summary_table["Missing N"]
```
